```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.tail()
```

```python
print(df.info())
print(df.describe())
```

```python
#handling age missing value with group by gender and passenger class and fill with median this is numerical data
df['age'] = df.groupby(['sex', 'pclass'])['age'].transform(lambda x: x.fillna(x.median()))
# handling data embark using mode operaion C = Cherbourg Q = Queenstown S = Southampton
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
# deck to missing value
df.drop('deck', axis=1, inplace=True)
# fill embarked_town missing value wtih unknown
df['embark_town'] = df['embark_town'].fillna('unknown')
```
